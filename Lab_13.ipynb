```python
#Task 1: Create Lambda Function for Fare Discount
discount_fare = lambda fare: fare - 200
print("Discounted fare :", discount_fare(2500))
```

```python
#Task 2: Apply map() Function for Fare Calculation
ticket_fare = [500, 1200, 2500, 800]
updated_fares = list(map(lambda fare: fare + 100,ticket_fare))
print(updated_fares)
```

```python
ticket_fares = [2500, 800, 3200, 600]

valid_fares = list(filter(lambda fare: fare > 1000, ticket_fares))
final_fares = list(map(lambda fare: fare + 100,valid_fares))
print(final_fares)
```

```python
passengers = ["Rahul", "Priya", "Amit"]
seats = ["S1", "S2", "S3"]
fares = [2500, 1800, 3200]

reservation_data = list(zip(passengers, seats, fares))
print("===================================")
print(" Railway Reservation Report ")
print("===================================")
for index, data in enumerate(reservation_data):
   passenger, seat, fare = data
   final_fare = (lambda x: x - 100)(fare)
   print("Reservation ID :", index + 1)
   print("Passenger :", passenger)
   print("Seat :", seat)
   print("Final Fare :", final_fare)
   print("-----------------------------------")
```
